## 1. Supervised vs. Unsupervised Learning Paradigms

Supervised learning maps inputs X to known labels Y (regression, classification). Unsupervised learning discovers latent structure, groupings, or representations without explicit labels (clustering, dimensionality reduction, self-supervised pre-training).

## 2. Loss Curves, Overfitting & Regularization (L1/L2)

Overfitting occurs when a model memorizes training noise instead of generalizable patterns. Regularization (L1 Lasso for sparsity, L2 Ridge for weight shrinking) penalizes model complexity.


````python
def simulate_training_epoch(epoch: int, overfit: bool = True):
    train_loss = 1.0 / (epoch + 1) + 0.05
    if overfit:
        val_loss = 1.0 / (epoch + 1) + (0.04 * max(0, epoch - 4)) + 0.08
    else:
        val_loss = 1.0 / (epoch + 1) + 0.12
    return round(train_loss, 4), round(val_loss, 4)

print("Epoch | Train Loss | Val Loss (Unregularized Overfitting)")
print("-" * 50)
for ep in range(1, 9):
    t_loss, v_loss = simulate_training_epoch(ep, overfit=True)
    status = "OVERFITTING" if v_loss > t_loss * 2 else "OK"
    print(f"  {ep:02d}  |   {t_loss:.4f}   |   {v_loss:.4f}  [{status}]")
````

## 3. K-Fold Cross-Validation

K-Fold Cross-Validation partitions data into K subsets, iteratively training on K-1 folds and validating on the remaining fold to assess generalization variance.


````python
def k_fold_split(dataset: list, k: int = 4):
    fold_size = len(dataset) // k
    splits = []
    for i in range(k):
        val_idx = list(range(i * fold_size, (i + 1) * fold_size))
        train_idx = [idx for idx in range(len(dataset)) if idx not in val_idx]
        splits.append({"fold": i + 1, "train_count": len(train_idx), "val_count": len(val_idx)})
    return splits

data_sample = list(range(100))
folds = k_fold_split(data_sample, k=5)
for f in folds:
    print(f"Fold {f['fold']}: Train samples={f['train_count']}, Val samples={f['val_count']}")
````

## Hands-On Capstone Lab: Build a Regularized Linear Classifier with K-Fold Validation

### Objective:
Implement an end-to-end binary classifier with L2 weight regularization and measure generalization performance.

### Instructions:
1. Synthesize a 2D linearly separable dataset.
2. Implement a sigmoid activation function.
3. Compute logits, predicted probability, and L2 complexity penalty.


````python
import math

def sigmoid(z: float) -> float:
    return 1.0 / (1.0 + math.exp(-max(min(z, 20.0), -20.0)))

weights = [0.5, -0.2]
bias = 0.1
l2_lambda = 0.01

test_sample = [1.5, 0.8]
raw_logit = sum(w * x for w, x in zip(weights, test_sample)) + bias
prob = sigmoid(raw_logit)
l2_penalty = l2_lambda * sum(w**2 for w in weights)

print(f"Input: {test_sample}")
print(f"Logit: {raw_logit:.4f} -> P(Class=1): {prob:.4f}")
print(f"L2 Complexity Penalty: {l2_penalty:.6f}")
````

### Question 2: How does L2 regularization (Ridge) penalize model parameters?
- **A)** By adding the sum of absolute weight values to the loss
- **B)** By adding the sum of squared weight values multiplied by lambda to the loss

<details>
<summary><b>View Answer & Explanation</b></summary>

**Correct Answer:** B) By adding the sum of squared weight values multiplied by lambda to the loss

**Explanation:** L2 regularization penalizes large weights by adding the sum of squared weights (lambda * sum(w^2)), shrinking parameter magnitudes.
